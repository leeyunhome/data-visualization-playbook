### Python으로 재현 (선택)

엑셀에서 만든 차트를 Python으로도 다시 그려보고 싶으면 아래에 이어간다.

> **데이터:** 애플 주가 CSV(`.gitignore` 처리, 로컬에만 있음). 강사님이 예로 든
> "가상 기업 3개 월간 매출"은 AI가 만든 더미 데이터라 재현할 원본이 없어, 같은 요지를
> 실측 데이터(연도별 종가)로 보여준다.

```python
import sys

sys.path.append("..")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from viz_utils import setup, load_sample

setup()
rng = np.random.default_rng(0)

df = pd.read_csv("../references/lecture_data/애플 주가/애플 주가.csv")
df["Date"] = pd.to_datetime(df["Date"])
year_end = df.set_index("Date")["Close"].resample("YE").last()
year_end.index = year_end.index.year
year_end
```

### 라인 차트 — 중간 과정이 다 보인다

```python
fig, ax = plt.subplots(figsize=(9, 4.3))
ax.plot(year_end.index, year_end.values, marker="o", color="#4C78A8")
ax.set(xlabel="연도", ylabel="종가($)", title="라인 차트 — 매년의 등락이 보인다")
plt.show()

ups = (year_end.diff().dropna() > 0).sum()
downs = (year_end.diff().dropna() < 0).sum()
print(f"연도별로 보면: 상승 {ups}년, 하락 {downs}년 — 오르락내리락했다")
```

### 슬로프 차트 — 양끝만 남기면

```python
fig, ax = plt.subplots(figsize=(4.5, 4.3))
first, last = year_end.index[0], year_end.index[-1]
ax.plot([0, 1], [year_end[first], year_end[last]], marker="o", color="#E45756", lw=2)
ax.set_xticks([0, 1], [first, last])
ax.set_ylabel("종가($)")
ax.set_title("슬로프 차트 — 계속 오른 것처럼 보인다")
plt.show()

print(f"{first} → {last}: {year_end[first]:.1f} → {year_end[last]:.1f} "
      f"({(year_end[last]/year_end[first]-1)*100:.0f}% 상승)")
```

**강의 요지 그대로 확인됨.** 슬로프 차트만 보면 10년간 꾸준히 오른 것 같지만,
라인 차트를 보면 2개 연도(2015, 2018)는 하락했다. "이런 중간 과정을 생략하고 결과만 보여줄 때
슬로프 차트를 활용한다"는 강의 설명이 정확하다 — **바꿔 말하면, 슬로프 차트는 그 자체로
정직하지 않은 게 아니라 무엇을 숨기고 있는지 아는 상태에서 써야 하는 차트다.**
