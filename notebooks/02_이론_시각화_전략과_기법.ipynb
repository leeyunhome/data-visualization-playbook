## 2-1. 기법을 고르는 순서

1. **무엇을 말하려는가** — 분포 / 비교 / 관계 / 구성 / 추세
2. **데이터의 차원은 몇 개인가** — 변수 개수와 각 변수의 타입
3. **변수 타입** — 범주형(명목·순서) / 수치형(이산·연속) / 시간
4. 위 셋이 정해지면 후보 차트는 2~3개로 줄어든다. 그중 **가장 단순한 것**을 고른다.

## 2-2. 데이터의 차원

| 차원 | 예시 | 대표 기법 |
| --- | --- | --- |
| 1차원 | 학생 300명의 점수 | 히스토그램, KDE, 박스플롯, 바이올린 |
| 2차원 | 공부시간 vs 점수 | 산점도, 막대, 히트맵 |
| 시계열 | 10분 간격 센서값 | 선그래프, 이동평균, 계절 분해 |
| 다차원(3~) | 6개 변수 x 180개 샘플 | 산점도 행렬, 평행좌표, 레이더, 히트맵 |
| 고차원 | 수십~수백 변수 | PCA, t-SNE, UMAP 등 차원 축소 |

- 내 데이터는 몇 차원인가?
